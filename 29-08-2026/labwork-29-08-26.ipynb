1. Fibonacci

```python
n = 5
a, b = 0, 1
for i in range(n + 1):
    print(a, end=" ")
    a, b = b, a + b
```

2. Multiplication Table

```python
n = int(input("Enter a number: "))
print("Multiplication table upto 10 for ", n, "is: ")
for i in range(1, 11):
    print(n, "X", i, "=", n * i)
```

3. Pyramid

```python
n = 5
for i in range(1, n + 1):
    space = n - i
    star = 2 * i - 1
    print(" " * space + "*" * star)
```

4. Right Angled Triangle

```python
n = 5
for i in range(n):
    for j in range(i + 1):
        print("*", end=" ")
    print()
```

5. Inverted Triangle

```python
n = 5
for i in range(n):
    for j in range(i, n):
        print("*", end=" ")
    print()
```
